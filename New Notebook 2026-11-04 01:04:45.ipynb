```python

print("Spark version:", spark.version)

data = [
    (101, "Customer_A", 0.92),
    (102, "Customer_B", 0.65),
    (103, "Customer_C", 0.21)
]

columns = ["account_id", "customer_name", "propensity_score"]

df = spark.createDataFrame(data, columns)

display(df)
```

```python

display(spark.sql("SHOW CATALOGS"))
```

```python
display(spark.sql("SHOW SCHEMAS"))
```

```python
display(spark.sql("SHOW TABLES"))
```

```python

# Replace with your actual workspace catalog
catalog_name = "workspace"

schema_name = "default"
table_name = "customer_propensity"

full_table_name = (
    f"`{catalog_name}`."
    f"`{schema_name}`."
    f"`{table_name}`"
)

print(full_table_name)
```

```python

df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(full_table_name)

print("Delta table created successfully!")
```

```python

saved_df = spark.table(full_table_name)

display(saved_df)
```

```python

display(
    spark.sql(
        f"DESCRIBE DETAIL {full_table_name}"
    )
)
```
